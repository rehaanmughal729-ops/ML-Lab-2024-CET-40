Task 08

```python
def findDog(sentence_text):
    return 'dog' in sentence_text.lower()
 
print(f"Q8: {findDog('Is there a dog here?')} ")
print(f"Q8: {findDog('My DOG is big')} ")
print(f"Q8: {findDog('Is there a cat here?')} ") 
```

Task 09

```python
def countDog(sentence_text):
    return sentence_text.lower().count('dog')
 
print(f"Q9: {countDog('This dog runs faster than the other dog dog')} ")
print(f"Q9: {countDog('Dog DOG dog')} ")
```

Task 10

```python
def caught_speeding(driving_speed, birthday_flag):
    speed_allowance = 5 if birthday_flag else 0
    if driving_speed <= 60 + speed_allowance:
        return 'No Ticket'
    elif driving_speed <= 80 + speed_allowance:
        return 'Small Ticket'
    else:
        return 'Big Ticket'
 
print(f"Q10: {caught_speeding(81, True)} ")  
print(f"Q10: {caught_speeding(81, False)} ")
print(f"Q10: {caught_speeding(65, True)} ")
```
